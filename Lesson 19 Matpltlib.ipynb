# Exercise 2: Customer Acquisition Comparison

```python
months = [1, 2, 3, 4, 5, 6]
campaign_a = [50, 65, 72, 90, 110, 125]
campaign_b = [45, 60, 68, 75, 95, 105]

plt.plot(months, campaign_a, '--k', label= 'Campaign A')
plt.plot(months, campaign_b, 'og', label= 'Campaign B')
plt.title('Marketing Campaign Trend Analysis')
plt.xlabel('Months')
plt.ylabel('Customers Acquired')
plt.xlim(1, 6)
plt.legend()
plt.show()

## Observation
# Both campaigns show an upward trend in customers acquired from Month 1 to Month 6.
# Campaign A acquires more customers than Campaign B in every month
# and the difference between them generally increases over the period.
```

# Exercise 3: Healthcare Performance Dashboard
### Management wants to understand how patient volume, average waiting time, and patient satisfaction have changed over six months.



```python
months = [1, 2, 3, 4, 5, 6]
patients = [420, 450, 480, 510, 495, 540]
waiting_time = [65, 62, 58, 55, 60, 49]
satisfaction = [72, 74, 76, 78, 77, 83]

fig1 = plt.figure(figsize = (12, 10))

patient_chart = fig1.add_subplot(3,1,1)
waiting_time_chart = fig1.add_subplot(3,1,2)
satisfaction_chart = fig1.add_subplot(3,1,3)

patient_chart.plot(months, patients)
waiting_time_chart.plot(months, waiting_time)
satisfaction_chart.plot(months, satisfaction)

patient_chart.set_title('Number Of Patients Treated')
waiting_time_chart.set_title('Average Waiting Time')
satisfaction_chart.set_title('Patient Satisfaction')

patient_chart.set_xlabel('Months')
waiting_time_chart.set_xlabel('Months')
satisfaction_chart.set_xlabel('Months')

patient_chart.set_ylabel('Total Number Patients')
waiting_time_chart.set_ylabel('Average Waiting Time (Minute)')
satisfaction_chart.set_ylabel('Patients Satisfaction (%)')

fig1.tight_layout()
plt.show()

# Report: Hospital Performance Analysis
# The charts show changes in patient numbers, average waiting time, and patient satisfaction over six months.
# Patients treated: The number increased from 420 in Month 1 to 540 in Month 6, although there was a small decline in Month 5.
# Average waiting time: Waiting time decreased from 65 minutes to 49 minutes, suggesting that patients spent less time waiting over the period.
# Patient satisfaction: Satisfaction increased from 72% to 83%, with a slight decline in Month 5.

```

# Bonus Challenge 

##### I would not plot all three indicators on the same y-axis because they measure different things and use different scales. The number of patients is measured in people, waiting time is measured in minutes, and satisfaction is measured in percentages. Putting them on the same scale could make some trends difficult to understand or misleading. It is better to use separate charts so each indicator is easy to read and interpret.

##  What happened to patient volume over the six months?

##### Patient volume generally increased from 420 patients in Month 1 to 540 in Month 6. There was a slight drop in Month 5, but the number increased again in the final month.

## How did average waiting time change?

##### Average waiting time decreased from 65 minutes in Month 1 to 49 minutes in Month 6. Although there was a small increase in Month 5, the overall trend shows that waiting time reduced.

## What happened to satisfaction in the final month?

##### Patient satisfaction increased from 77% in Month 5 to 83% in Month 6, which was the highest figure recorded during the six months.

##  Do the patterns prove that shorter waiting times caused higher satisfaction? Why or why not?

##### No, they do not prove that shorter waiting times caused higher satisfaction. The charts show that waiting time decreased while satisfaction generally increased, but other factors could have contributed to the improvement. For example, better staff attitude, improved facilities, or better treatment could also have influenced how patients felt.

##  What additional data would help the hospital investigate the relationship between waiting time and satisfaction?

##### I would need individual patient waiting times and their satisfaction ratings to see whether patients who waited less were generally more satisfied. I would also look at staffing levels, patient numbers per day, the quality of care, and patient feedback to understand what else might be influencing satisfaction.

##### Conclusion: The hospital appears to be treating more patients while reducing waiting times and improving satisfaction. This is encouraging, but more information is needed to understand the reasons behind these changes and determine whether they are connected.

# Mini Project: Business Performance Visualization
### Project title: Visualizing Business Performance with Matplotlib

```python
months = [1, 2, 3, 4, 5, 6]
revenue = [150000, 165000, 158000, 190000, 215000, 240000]
customers = [80, 92, 88, 110, 125, 140]
product_a = [45, 50, 48, 60, 68, 75]
product_b = [35, 40, 42, 45, 52, 58]

fig1 = plt.figure(figsize=(12, 10))

revenue_chart = fig1.add_subplot(4,1,1)
customers_chart = fig1.add_subplot(4,1,2)
product_chart = fig1.add_subplot(4,1,3)

revenue_chart.plot(months, revenue, 'r--o')
customers_chart.plot(months, customers, 'g--o' )
product_chart.plot(months, product_a, 'r-o', label= 'Product A')
product_chart.plot(months, product_b, 'y--s',  label= 'Product B')

revenue_chart.set_title('Monthly revenue trend')
customers_chart.set_title('Customer growth')
product_chart.set_title('Product Comparison A and B')

revenue_chart.set_xlabel('Months')
customers_chart.set_xlabel('Months')
product_chart.set_xlabel('Months')

revenue_chart.set_ylabel('Revenue (₦)')
customers_chart.set_ylabel('Number Of Customers')
product_chart.set_ylabel('Numbers Of Products')



fig1.tight_layout()
product_chart.legend()
plt.show()
```

```python
# Business Performance Report

# Revenue generally increased over the six months, rising from ₦150,000 in Month 1 to ₦240,000 in Month 6. 
# However, revenue declined slightly in Month 3 before increasing again in Month 4.
# Customer numbers followed a similar pattern. They increased from 80 in Month 1 to 140 in Month 6, although there was a small decline in Month 3.
# Product A recorded 75 units sold in Month 6, compared with 58 units for Product B. Product A therefore had higher sales in the final month.
# Overall, the business recorded growth in revenue, customer numbers, and sales of both products. 
#However, more information about product prices, costs, marketing activities, and profit margins would help explain what drove the growth and which product contributes more to profit.
```
